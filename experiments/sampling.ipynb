```python
import numpy as np

from colorpalettes.color import Color

colors: list[Color] = []
for _ in range(5_000):
    rgb = np.random.uniform(0.0, 1.0, size=3)
    colors.append(Color(tuple(rgb)))
```

```python
%matplotlib ipympl
```

```python
from matplotlib import pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(6, 6))
ax: Axes3D = fig.add_subplot(projection="3d")  # type: ignore

ax.scatter(
    [c.JCh[0] for c in colors],
    [c.JCh[1] for c in colors],
    [c.JCh[2] for c in colors],  # type: ignore
    c=[c.rgb for c in colors],
    marker=".",
)
ax.set_xlabel("Lightness")
ax.set_ylabel("Chroma")
ax.set_zlabel("Hue")
```
